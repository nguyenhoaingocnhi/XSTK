```python
import numpy as np
from scipy.stats import norm

# 1. Khai báo thông số từ đề bài Ví dụ 5.3 trang 105 SGK
n = 25
x_tb = 50.1
σ = 1.0
α = 1 - 0.95  # Độ tin cậy 95% -> α = 0.05

# 2. Tính giá trị tới hạn z_(α/2) và sai số ước lượng
z_α_2 = norm.ppf(1 - α / 2)
ε = z_α_2 * (σ / np.sqrt(n))

# 3. In kết quả
print("--- KẾT QUẢ VÍ DỤ 5.3 (trang 105 SGK) ---")
print(f"z_(α/2) = z_0.025     : {z_α_2:.2f}")
print(f"Sai số ước lượng của mẫu  : {ε:.1f}")
print(f"Khoảng tin cậy 95% của μ  : {x_tb - ε:.1f} < μ < {x_tb + ε:.1f}")
```

```python
import numpy as np
from scipy.stats import t

# 1. Khai báo thông số từ đề bài Ví dụ 5.4 trang 107 SGK
n = 16
x_tb = 24308
s_sample = 727
α = 1 - 0.95  # Độ tin cậy 95% -> α = 0.05

# 2. Tính giá trị tới hạn t_(n-1; α/2) và sai số ước lượng
t_α_2 = t.ppf(1 - α / 2, df=n - 1)
ε = t_α_2 * (s_sample / np.sqrt(n))

# 3. In kết quả
print("--- KẾT QUẢ VÍ DỤ 5.4 (trang 107 SGK) ---")
print(f"t_(15; 0.025)             : {t_α_2:.4f}")
print(f"Sai số ước lượng của mẫu  : {ε:.0f}")
print(f"Khoảng tin cậy 95% của μ  : {x_tb - ε:.0f} < μ < {x_tb + ε:.0f}")
```
