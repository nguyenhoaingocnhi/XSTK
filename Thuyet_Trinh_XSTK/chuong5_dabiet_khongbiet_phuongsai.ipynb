```python
import numpy as np
from scipy.stats import norm

# 1. Khai báo thông số từ đề bài Ví dụ 5.3
n = 25
x_tb = 50.1
sigma = 1.0
alpha = 1 - 0.95  # Độ tin cậy 95% -> alpha = 0.05

# 2. Tính giá trị tới hạn z_(alpha/2) và sai số ước lượng
z_alpha_2 = norm.ppf(1 - alpha / 2)
sai_so = z_alpha_2 * (sigma / np.sqrt(n))

# 3. In kết quả
print("--- KẾT QUẢ VÍ DỤ 5.3 ---")
print(f"z_(alpha/2) = z_0.025     : {z_alpha_2:.2f}")
print(f"Sai số                    : {sai_so:.1f}")
print(f"Khoảng tin cậy 95% của mu : {x_tb - sai_so:.1f} < mu < {x_tb + sai_so:.1f}")
```

```python
import numpy as np
from scipy.stats import t

# 1. Khai báo thông số từ đề bài Ví dụ 5.4
n = 16
x_tb = 24308
sigma = 727
alpha = 1 - 0.95  # Độ tin cậy 95% -> alpha = 0.05

# 2. Tính giá trị tới hạn t_(n-1; alpha/2) và sai số ước lượng
t_alpha_2 = t.ppf(1 - alpha / 2, df=n - 1)
sai_so = t_alpha_2 * (sigma / np.sqrt(n))

# 3. In kết quả
print("--- KẾT QUẢ VÍ DỤ 5.4 ---")
print(f"t_(15; 0.025)             : {t_alpha_2:.4f}")
print(f"Sai số                    : {sai_so:.0f}")
print(f"Khoảng tin cậy 95% của mu : {x_tb - sai_so:.0f} < mu < {x_tb + sai_so:.0f}")
```
