```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.stats import norm
```

```python
# 1. Khai báo số liệu (dữ liệu từ Ví dụ 5.6 trang 109 SGK)
data_goc = np.array([
    1.23, 0.49, 0.49, 1.08, 0.59, 0.28, 0.18,
    0.10, 0.94, 1.33, 0.19, 1.16, 0.98, 0.34,
    0.34, 0.19, 0.21, 0.40, 0.04, 0.83, 0.05,
    0.63, 0.34, 0.75, 0.04, 0.86, 0.43, 0.04,
    0.81, 0.15, 0.56, 0.84, 0.87, 0.49, 0.52,
    0.25, 1.20, 0.71, 0.19, 0.41, 0.50, 0.56,
    1.10, 0.65, 0.27, 0.27, 0.50, 0.77, 0.73,
    0.34, 0.17, 0.16, 0.27
])
# In ra bảng số liệu ban đầu
print("--- BẢNG SỐ LIỆU BAN ĐẦU (Ví dụ 5.6 trang 109 SGK) ---")
df_ban_dau = pd.DataFrame({'STT': np.arange(1, len(data_goc) + 1), 'Giá trị': data_goc})
print(df_ban_dau.to_string(index=False))
print("\n")
```

```python
# 2. Sắp xếp số liệu tăng dần
x_j = np.sort(data_goc)
# In ra bảng sau khi sắp xếp
print("--- BẢNG SỐ LIỆU SAU SẮP XẾP ---")
df_sap_xep = pd.DataFrame({'STT': np.arange(1, len(x_j) + 1), 'Giá trị tăng dần': x_j})
print(df_sap_xep.to_string(index=False))
print("\n")
```

```python
# 3. Tính toán các đại lượng đặc trưng
n = len(x_j)
x̄ = np.mean(x_j)
s = np.std(x_j, ddof=1)

print(f"--- KẾT QUẢ VÍ DỤ 5.6 (trang 109 SGK) ---")
print(f"Cỡ mẫu (n)                : {n}")
print(f"Trung bình mẫu (x̄)        : {x̄:.2f}(10⁻⁴%)")
print(f"Độ lệch chuẩn mẫu (s)     : {s:.2f}(10⁻⁴%)")

# Tính toán các thông số
j = np.arange(1, n + 1)
p_j = (j - 0.5) / n
# p = norm.cdf(p_j)
z_j = norm.ppf(p_j)

# In bảng tính toán
print("--- BẢNG TÍNH TOÁN ---")
df = pd.DataFrame({
    'j': j,
    'x_j': x_j,
    'p_j = (j-0.5)/n': p_j,
    'z_j = Φ⁻¹(p_j)': z_j
})
print(df.to_string(index=False, float_format="%.3f"))
print("\n")
```

```python
# 4. Vẽ biểu đồ xác suất chuẩn
plt.figure(figsize=(12, 5))

# Vẽ các điểm dữ liệu thực tế
plt.scatter(z_j, x_j, color='green', zorder=5, label='Dữ liệu thực tế (x_j)')

# Vẽ đường thẳng: x = s*z + x̄
z_line = np.linspace(min(z_j) - 0.2, max(z_j) + 0.2)
x_line = s * z_line + x̄
plt.plot(z_line, x_line, color='blue', linestyle='--', label='x = s*z + x̄')

plt.title('Biểu đồ xác suất chuẩn')
plt.xlabel('z_j')
plt.ylabel('x_j')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()
```
