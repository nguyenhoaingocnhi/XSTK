```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.stats import norm

# Tạo dữ liệu lệch phải (phân phối exponential)
np.random.seed(42)
non_normal_data = np.random.exponential(scale=2, size=1000)

# Tính trung bình và độ lệch chuẩn
mean = np.mean(non_normal_data)
std = np.std(non_normal_data)

# Tạo DataFrame
df_non_normal = pd.DataFrame(non_normal_data, columns=['Giá trị'])

# Tạo các giá trị x để vẽ đường cong chuẩn
x = np.linspace(min(non_normal_data), max(non_normal_data), 1000)
y = norm.pdf(x, mean, std)

# Vẽ biểu đồ hình chuông
plt.figure(figsize=(10, 6))
plt.hist(non_normal_data, bins=30, density=True, color='lightblue', edgecolor='black', alpha=0.7, label='Tần suất quan sát')
plt.plot(x, y, 'r--', linewidth=2, label='Phân phối chuẩn (tham chiếu)')
plt.axvline(mean, color='black', linestyle='--', label=f'Kỳ vọng E(X) = {mean:.2f}')
plt.title('Biểu đồ hình chuông của dữ liệu không chuẩn (Lệch phải)')
plt.xlabel('Giá trị')
plt.ylabel('Mật độ xác suất')
plt.legend()
plt.grid(True)
plt.show()
```
